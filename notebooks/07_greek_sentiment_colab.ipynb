## Αξιολόγηση σε πραγματικές ελληνικές κριτικές

```python
from pathlib import Path
hotel_test = Path('greek_hotel_test.csv')
if hotel_test.is_file():
    from reviewnlp.greek.evaluate import evaluate_hotel_csv
    report = evaluate_hotel_csv('runs/greek_bert', hotel_test, 'runs/greek_hotel_evaluation')
    print(report['test'])
else:
    print('Απαιτείται επισημειωμένο greek_hotel_test.csv για αξιολόγηση στο ξενοδοχειακό πεδίο.')
```
